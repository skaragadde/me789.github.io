```python
import numpy as np

def lagrange_interpolation(x_data, y_data, x):
    """
    Evaluates the Lagrange interpolating polynomial at the target point x.
    
    x_data: Known x coordinates
    y_data: Known y coordinates
    x: Target point to interpolate
    """
    x_data = np.array(x_data, dtype=float)
    y_data = np.array(y_data, dtype=float)
    n = len(x_data)
    total = 0.0

    for i in range(n):
        # Calculate basis polynomial L_i(x)
        term = y_data[i]
        for j in range(n):
            if i != j:
                term *= (x - x_data[j]) / (x_data[i] - x_data[j])
        total += term

    return total
```

```python

# Example usage
x_pts = [0, 1, 2, 5]
y_pts = [2, 3, 12, 147]

target_x = 3
result = lagrange_interpolation(x_pts, y_pts, target_x)
print(f"Interpolated y value at x = {target_x}: {result}")
```
